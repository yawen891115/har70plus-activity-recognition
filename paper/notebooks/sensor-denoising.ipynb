```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 降噪504（体现噪声）
# 求504人back传感器的合向量的模
# 导入504df
df_5041 = pd.read_csv(f'/Users/zhengyawen/Desktop/zywhar70/har70dataset/504.csv')  # 请将路径替换为实际路径

# 计算欧氏距离的函数
def euclidean_distance(row):
    return np.sqrt(row['back_x']**2 + row['back_y']**2 + row['back_z']**2)

# 应用函数并将结果添加为新列
df_5041['euclidean_distance'] = df_5041.apply(euclidean_distance, axis=1)
print(df_5041)

# 对于每个label，绘制只包含线条的时间序列图
for label in [1, 3, 4, 5, 6, 7, 8]:
    # 筛选出指定 label 的数据子集
    subset = df_5041[df_5041['label'] == label]
    # 为每个子集创建一个整数序列
    x = range(1, len(subset) + 1)
    plt.figure(figsize=(10, 6))
    # 绘制线条，不显示点，线条更细
    plt.plot(x, subset['euclidean_distance'], linestyle='-', linewidth=0.5)
    plt.title(f'Time Series Plot for Label {label}')
    plt.xlabel('Integer Sequence')
    plt.ylabel('Euclidean Distance for 504')
    plt.grid(True)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
#由上面的合向量图看出来有噪声
#用均值滤波去噪，按照每列为总数据集merged_df去噪
# 创建数据副本
df_504 = df_5041.copy()
N = 5
# 指定要过滤的加速度数据列
columns_to_filter = ['euclidean_distance']
# 对指定的列应用移动均值滤波
for col in columns_to_filter:
    df_504[col + '_filtered'] = df_504[col].rolling(window=N).mean()
df_504= df_504.drop(columns=['back_x', 'back_y', 'back_z', 'thigh_x', 'thigh_y', 'thigh_z'])
#对于每个label，我都画出对应的时间戳和欧氏距离的time series plot
# 循环遍历每个 label
for label in [1, 3, 4, 5, 6, 7, 8]:
    # 筛选出指定 label 的数据子集
    subset = df_504[df_504['label'] == label]
    # 为每个子集创建一个整数序列
    x = range(1, len(subset) + 1)
    plt.figure(figsize=(10, 6))
    # 绘制线条，不显示点，线条更细
    plt.plot(x, subset['euclidean_distance_filtered'], linestyle='-', linewidth=0.5)
    plt.title(f'Time Series Plot for Label {label}')
    plt.xlabel('Integer Sequence')
    plt.ylabel('Euclidean Distance for 504')
    plt.grid(True)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 导入数据
df_5041 = pd.read_csv('/Users/zhengyawen/Desktop/zywhar70/har70dataset/504.csv')

# 计算欧氏距离的函数
def euclidean_distance(row):
    return np.sqrt(row['back_x']**2 + row['back_y']**2 + row['back_z']**2)

# 应用函数并将结果添加为新列
df_5041['euclidean_distance'] = df_5041.apply(euclidean_distance, axis=1)

# 用均值滤波去噪
N = 5  # 滤波器窗口大小
df_504 = df_5041.copy()
df_504['euclidean_distance_filtered'] = df_504['euclidean_distance'].rolling(window=N).mean()

# 删除不必要的列
df_504 = df_504.drop(columns=['back_x', 'back_y', 'back_z', 'thigh_x', 'thigh_y', 'thigh_z'])

# 确定全局y轴范围
y_min = min(df_504['euclidean_distance'].min(), df_504['euclidean_distance_filtered'].min())
y_max = max(df_504['euclidean_distance'].max(), df_504['euclidean_distance_filtered'].max())

# 对于每个label，绘制降噪前和降噪后的时间序列图
for label in [1, 3, 4, 5, 6, 7, 8]:
    subset = df_504[df_504['label'] == label]
    x = range(1, len(subset) + 1)
    
    plt.figure(figsize=(10, 6))
    
    # 降噪前
    plt.plot(x, subset['euclidean_distance'], linestyle='-', linewidth=0.5, label='Before Noise Reduction')
    # 降噪后
    plt.plot(x, subset['euclidean_distance_filtered'], linestyle='-', linewidth=0.5, label='After Noise Reduction')
    
    plt.title(f'Time Series Plot for Label {label}')
    plt.xlabel('Integer Sequence')
    plt.ylabel('Euclidean Distance for 504')
    plt.ylim([y_min, y_max])  # 设置y轴范围
    plt.legend()
    plt.grid(True)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 导入数据
df_5121 = pd.read_csv('/Users/zhengyawen/Desktop/zywhar70/har70dataset/512.csv')

# 计算欧氏距离的函数
def euclidean_distance(row):
    return np.sqrt(row['thigh_x']**2 + row['thigh_y']**2 + row['thigh_z']**2)

# 应用函数并将结果添加为新列
df_5121['euclidean_distance'] = df_5121.apply(euclidean_distance, axis=1)

# 用均值滤波去噪
N = 5  # 滤波器窗口大小
df_512 = df_5121.copy()
df_512['euclidean_distance_filtered'] = df_512['euclidean_distance'].rolling(window=N).mean()

# 删除不必要的列
df_512 = df_512.drop(columns=['back_x', 'back_y', 'back_z', 'thigh_x', 'thigh_y', 'thigh_z'])

# 确定全局y轴范围
y_min = min(df_512['euclidean_distance'].min(), df_512['euclidean_distance_filtered'].min())
y_max = max(df_512['euclidean_distance'].max(), df_512['euclidean_distance_filtered'].max())

# 对于每个label，绘制降噪前和降噪后的时间序列图
for label in [1, 3, 4, 5, 6, 7, 8]:
    subset = df_512[df_512['label'] == label]
    x = range(1, len(subset) + 1)
    
    plt.figure(figsize=(10, 6))
    
    # 降噪前
    plt.plot(x, subset['euclidean_distance'], linestyle='-', linewidth=0.5, label='Before Noise Reduction',color='black')
    # 降噪后
    plt.plot(x, subset['euclidean_distance_filtered'], linestyle='-', linewidth=0.5, label='After Noise Reduction',color='red')
    
    plt.title(f'Time Series Plot for Label {label}')
    plt.xlabel('Integer Sequence')
    plt.ylabel('Euclidean Distance for 512')
    plt.ylim([y_min, y_max])  # 设置y轴范围
    plt.legend()
    plt.grid(True)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
########################################
#import some nessisary package
########################################
import pandas as pd
import numpy as np
#visualisation
import matplotlib.pyplot as plt
import seaborn as sns
# data preprocessing
from sklearn.preprocessing import StandardScaler
#import evaluation index such as confusion matrix, accuracy_score, f1_score, precision_score and recall_score
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix
# data splitting
from sklearn.model_selection import train_test_split
#import important machine learning models
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.experimental import enable_hist_gradient_boosting  # noqa
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis
from sklearn.neural_network import MLPClassifier



###########################################
#import the csv into jupyter notebook
#the dataset
############################################
merged_df1 = pd.DataFrame()
# 循环读取并合并18个数据集
for i in range(501, 503):
    # 读取数据集
    df = pd.read_csv(f'/Users/zhengyawen/Desktop/zywhar70/har70dataset/{i}.csv')
    # 第一个数据集直接赋值给merged_df
    if i == 501:
        merged_df1 = df
    else:
        # 根据共同列进行合并
        merged_df1 = pd.concat([merged_df1,df],ignore_index=True)
# 输出合并后的数据集
print(merged_df1)



###########################
#data preprocessing
###########################
#check whether the dataset exist missing value
# 查看每列缺失值的数量
missing_values = merged_df1.isnull().sum()
# 查看整个数据集缺失值的总数量
total_missing = merged_df1.isnull().sum().sum()
# 打印缺失值情况
print(missing_values)
print(total_missing)

#降噪
# 创建数据副本
merged_df = merged_df1.copy()
# 设置移动均值滤波的窗口大小
N = 5
# 指定要过滤的加速度数据列
columns_to_filter = ['back_x', 'back_y', 'back_z', 'thigh_x', 'thigh_y', 'thigh_z']
# 对指定的列应用移动均值滤波，并直接更新这些列
for col in columns_to_filter:
    merged_df[col] = merged_df[col].rolling(window=N).mean()
# 删除移动均值滤波导致的前N-1行空值
merged_df = merged_df.iloc[N-1:].reset_index(drop=True)


##################################
#统计学分析
##################################
#共线性检验
# 使用corr()函数计算相关系数矩阵
correlation_matrix = merged_df.drop(columns=['label']).corr()
# 打印相关系数矩阵
print(correlation_matrix)
# 使用Seaborn的heatmap函数绘制热力图
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
# 设置热力图的标题
plt.title('Correlation Heatmap')
# 显示图形
plt.show()

#共线性检验
#使用VIF（方差膨胀因子）检验共线性
from statsmodels.stats.outliers_influence import variance_inflation_factor
# 添加一个截距列，否则VIF求不出来，VIF是基于线性回归模型的
X_vif = merged_df.drop(columns=['timestamp','label'])
X_vif['Intercept'] = 1
# 初始化一个DataFrame来存储VIF结果
vif_data = pd.DataFrame()
vif_data["Variable"] = X_vif.columns
# 计算每个变量的VIF
vif_data["VIF"] = [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])]
# 打印结果
print(vif_data)
```
